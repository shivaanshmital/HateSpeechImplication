```python
dict_of_mappings = {'train':'train','test':'test','validation':'val'}
```

```python
final_data = {'test':[],'validation':[],'train':[]}
ids_rationale_not_found = []
```

```python
for train_test_val in dataset.keys():
  for data in dataset[train_test_val]:
    id = data['HITId']
    try:
      data['rationale'] = full_data[dict_of_mappings[train_test_val]][id]['rationale']
      final_data[train_test_val].append(data)
    except:
      ids_rationale_not_found.append(id)

```
